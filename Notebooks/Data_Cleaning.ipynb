```python
import cv2
import os

def clean_existing_dataset(source_folder, output_folder, size=(200, 200)):
    face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
    
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    files = [f for f in os.listdir(source_folder) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
    print(f"Starting cleanup of {len(files)} images from {source_folder}...")

    count = 0
    for filename in files:
        img_path = os.path.join(source_folder, filename)
        img = cv2.imread(img_path)
        if img is None: continue
            
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        
        # --- CHANGE: REMOVED cv2.equalizeHist HERE ---
        # We want raw, clean crops. Lighting fix happens during training.
        
        # Detect faces
        faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5)
        
        if len(faces) > 0:
            # Pick the largest face in the image
            x, y, w, h = max(faces, key=lambda rect: rect[2] * rect[3])
            face_crop = gray[y:y+h, x:x+w]
            
            # Resize
            face_final = cv2.resize(face_crop, size)
            
            save_path = os.path.join(output_folder, filename)
            cv2.imwrite(save_path, face_final)
            count += 1
        else:
            print(f"Warning: No face detected in {filename}.")
            
    print(f"Finished. Saved {count} faces to {output_folder}.")

# Define your paths here
source_train = r'C:\Users\hp\Desktop\Attendance-System-Using-Face-Recognition\Dataset\training\Training_RAW'
dest_train = r'C:\Users\hp\Desktop\Attendance-System-Using-Face-Recognition\Dataset\training\Cleaned_Training'

source_test = r'C:\Users\hp\Desktop\Attendance-System-Using-Face-Recognition\Dataset\test\Test_RAW'
dest_test = r'C:\Users\hp\Desktop\Attendance-System-Using-Face-Recognition\Dataset\test\Cleaned_test'

# Run cleanup
clean_existing_dataset(source_train, dest_train)
clean_existing_dataset(source_test, dest_test)
```
