```python
import cv2
import os

def clean_existing_dataset(source_folder, output_folder, size=(200, 200)):
    # 1. Load the detector (Corrected spelling: haarcascade)
    face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
    
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    # 2. Get list of all files in the source folder
    files = [f for f in os.listdir(source_folder) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
    
    print(f"Starting cleanup of {len(files)} images...")

    for filename in files:
        img_path = os.path.join(source_folder, filename)
        img = cv2.imread(img_path)
        
        if img is None:
            continue
            
        # 3. Convert to Grayscale
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        
        # 4. Detect Faces (minNeighbors=5 to ensure quality)
        faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5)
        
        if len(faces) > 0:
            # 5. Take the largest face in the photo
            x, y, w, h = max(faces, key=lambda rect: rect[2] * rect[3])
            
            # Crop to the face
            face_crop = gray[y:y+h, x:x+w]
            
            # 6. Resize to standard size (200x200)
            face_final = cv2.resize(face_crop, size)
            
            # 7. Save with the same name
            save_path = os.path.join(output_folder, filename)
            cv2.imwrite(save_path, face_final)
            print(f"Successfully cleaned: {filename}")
        else:
            print(f"Warning: No face detected in {filename}. Image skipped.")

# --- EXECUTION ---
# Use raw strings (r'') to avoid the path errors from before
source = r'C:\Users\hp\Desktop\Attendance-System-Using-Face-Recognition\Dataset\training\Training_RAW'
destination = r'C:\Users\hp\Desktop\Attendance-System-Using-Face-Recognition\Dataset\training\Cleaned_Training'

clean_existing_dataset(source, destination)
print("\nDone! All cleaned faces are in the Cleaned_Train folder.")

source1 = r'C:\Users\hp\Desktop\Attendance-System-Using-Face-Recognition\Dataset\test\Test_RAW'
destination1= r'C:\Users\hp\Desktop\Attendance-System-Using-Face-Recognition\Dataset\test\Cleaned_test'

clean_existing_dataset(source1, destination1)
print("\nDone! All cleaned faces are in the Cleaned_test folder.")
```
